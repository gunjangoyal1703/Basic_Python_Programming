```python
# Python Tuples
VeggiesList = ("Brocolli", "Potato", "Carrot", "Cauliflower", "Pumpkin")
print("Veggies:",(VeggiesList))
print(VeggiesList[1])
```

```python
VeggiesList = ("Brocolli", "Potato", "Carrot", "Cauliflower", "Pumpkin")
(VeggiesList[1]) = "Cabbage"
print(VeggiesList) # Tuples cannot be modified like lists
```

```python
#Append Items
VeggiesList = ("Brocolli", "Potato", "Carrot", "Cauliflower", "Pumpkin")
VeggiesList.append("Cabbage")
print(VeggiesList) # Tuples cannot be modified like lists
```

```python
# Add Tuple to Tuple
VeggiesList = ("Brocolli", "Potato", "Carrot", "Cauliflower", "Pumpkin")
y = ("Cabbage",)
VeggiesList += y
print(VeggiesList)
```

```python
# Tuple Slicing

# Example1
List1 = (20,30,50,60,40,10,70)
print(List1[1:3])
print(List1[1:3:1])
print(List1[1:5:2])
print(List1[1:4:3])
print(List1[4:6:-1]) # Goes one step back
print(List1[6:4:-1])
print(List1[6:4:-2])
print(List1[-2:-4:-2])

# Example2
Numbers = (10, 20, 30, 40, 50)
print("Elements from index 1 to 3:") ; print(Numbers[1:4])
print("First three elements:") ; print(Numbers[:4])
print("Elements from index 2 onwards:") ; print(Numbers[-3:])
print("All elements:") ; print(Numbers[:])
```

```python
# Various Marks
Marks = (78, 85, 90, 66, 88)
print("Maximum Marks:",max(Marks))
print("Minimum Marks:",min(Marks))
print("Total Marks:",sum(Marks))
print("Number of Students:",len(Marks))
print("Average Marks:",sum(Marks)/len(Marks))
print("Sort Marks in Ascending Order:",Marks.sort())
print(Marks) # Sort does not work in Tuples, as it is changing original input
```

```python
# Temperature Analysis
PuneTemp = (34, 32 ,37, 43, 23, 34, 43, 25, 36,38)
print("Pune's Temperature in first 5 days:",(PuneTemp)[:5])
print("Pune's Temperature in last 5 days:",(PuneTemp)[5:])
print("Pune's Temperature:",(PuneTemp))
print("Highest Temperature:",max(PuneTemp))
print("Lowest Temperature:",min(PuneTemp))
print("Average Temperature:",sum(PuneTemp)/len(PuneTemp))
```

```python
# Daily Expense Tracker
ExpenseTracker = (345, 76, 7564, 756, 674, 576, 566, 678, 678, 465)
print("Expenses:",(ExpenseTracker))
print("Total Expenses:",sum(ExpenseTracker))
print("Average Expenses:",sum(ExpenseTracker)/len(ExpenseTracker))
print("Highest Spending:",max(ExpenseTracker))
print("Lowest Spending:",min(ExpenseTracker))
```

```python
# Unpack Tuples
Result = ( "EDA", 88, "B")
Subject, Marks, Grade = Result
print(Subject)
print(Marks)
print(Grade)
if Marks >= 75:
  print("Distinction") # The space befor print is indentation, automatically given by python
```
