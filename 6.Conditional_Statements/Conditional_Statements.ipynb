```python
# Check if a Date is Valid and Return Incremented Date (Shorter Code)
# Input date in dd-mm-yyyy format
date = input("Enter date (dd-mm-yyyy): ")

# Split the date
parts = date.split("-")

# Check correct format
if len(parts) != 3:
    print("Invalid date")
    exit()

day = int(parts[0])
month = int(parts[1])
year = int(parts[2])

# Check leap year
leap = False
if (year % 400 == 0) or (year % 4 == 0 and year % 100 != 0):
    leap = True

# Find max days in the month
if month == 1 or month == 3 or month == 5 or month == 7 or month == 8 or month == 10 or month == 12:
    max_days = 31
elif month == 4 or month == 6 or month == 9 or month == 11:
    max_days = 30
elif month == 2:
    if leap:
        max_days = 29
    else:
        max_days = 28
else:
    print("Invalid date")
    exit()

# Validate day
if day < 1 or day > max_days:
    print("Invalid date")
    exit()

# Increment date
day += 1

if day > max_days:
    day = 1
    month += 1

    if month > 12:
        month = 1
        year += 1

# Print incremented date
print("Incremented date:", f"{day:02d}-{month:02d}-{year}")
```

```python
''' Problem Statement:
Calculate the gross salary of an employee based on basic salary
* if basic<=10000: HRA=20%, DA=80%
* if basic<=20000: HRA=25%, DA=90%
* otherwise HRA=30%, DA=95% '''
Basic = int(input("Enter the Basic Salary: "))
if Basic<=10000:
  HRA = 0.2*Basic
  DA = 0.8*Basic
  Gross = Basic + HRA + DA
  print("The Gross Salary is",Gross)
elif Basic<=20000:
   HRA = 0.25*Basic
   DA = 0.9*Basic
   Gross = Basic + HRA + DA
   print("The Gross Salary is",Gross)
else:
  HRA = 0.3*Basic
  DA = 0.95*Basic
  Gross = Basic + HRA + DA
  print("The Gross Salary is",Gross)
```

```python
''' Problem Statement:
Calculate Income Tax based on Annual Income
* if income<=2,50,000: No Tax
* iv income<=5,00,000: Tax=5%
* if income<=1000000: 20%
* above 1000000: 30% '''
income = int(input("Enter the Annual Income: "))
if income<=250000:
  print("No Tax")
elif income<=500000:
  tax=(income-250000)*0.05
  print("Tax Amount =",tax)
elif income<=1000000:
  tax=(income-250000)*0.2
  print("Tax Amount =",tax)
else:
  tax=(income-250000)*0.3
  print("Tax Amount =",tax)
```
